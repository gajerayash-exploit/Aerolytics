```python
from ultralytics import YOLO

# 1. Load the base AI Engine
print("Loading Base AI Engine for Agriculture...")
agri_model = YOLO('yolov8n.pt')

# 2. Train the model on the Crop Disease dataset
print("Starting Crop Health Training...")
agri_results = agri_model.train(
    data='data.yaml',    # This will safely read the agriculture yaml in this new folder
    epochs=15,           
    imgsz=640,
    project='runs/detect',    
    name='agri_model',   # Saves it specifically as 'agri_model'
    plots=True           
)

print("Agriculture Training Complete! Your second DaaS application is ready.")
```
